## 2.1 Dasar Teori Percabangan
Percabangan adalah struktur kontrol yang memungkinkan program mengambil keputusan berdasarkan kondisi tertentu. Dengan percabangan, program dapat mengeksekusi blok kode yang berbeda tergantung pada apakah suatu 
kondisi bernilai True atau False.

### 2.1.1 Struktur if
Struktur if adalah bentuk percabangan paling dasar. Blok kode di dalam if hanya dieksekusi jika kondisi bernilai True. Jika kondisi False, blok tersebut dilewati dan tidak ada aksi alternatif.

### 2.1.2 Struktur if-else
Struktur if-else menyediakan dua jalur eksekusi. Jika kondisi True, blok if dieksekusi. Jika False, blok else yang dieksekusi. Salah satu dari keduanya selalu dieksekusi.

### 2.1.3 Struktur if-elif-else
Struktur if-elif-else digunakan untuk menangani lebih dari dua kondisi. elif adalah singkatan dari 'else if'. Python mengevaluasi kondisi secara berurutan dari atas ke bawah, dan mengeksekusi blok pertama yang kondisinya True. Setelah satu blok dieksekusi, kondisi lain tidak dievaluasi lagi.

### 2.1.4 Multiple if Independen
Multiple if independen adalah penggunaan beberapa pernyataan if terpisah. Berbeda dengan elif, setiap if selalu dievaluasi terlepas dari hasil kondisi sebelumnya. Ini berguna ketika lebih dari satu kondisi bisa terpenuhi sekaligus (misalnya: akumulasi diskon).

### 2.1.5 Operator Logika
Operator logika digunakan untuk menggabungkan beberapa kondisi dalam satu pernyataan if. Terdapat tiga operator utama: AND (kedua kondisi harus True), OR (cukup satu kondisi True), dan NOT (membalik nilai boolean).

### 2.1.6 Match Case 
Pernyataan match-case adalah fitur structural pattern matching yang diperkenalkan pada Python 3.10. Ia menyediakan sintaks yang lebih bersih dibanding if-elif panjang untuk pencocokan nilai diskrit. case _ berfungsi sebagai wildcard (setara dengan else) yang menangani semua nilai yang tidak cocok.

## 2.2 Kode Program

### 2.2.1 Program "if1_2511533011"

```python
ipk_3011 = float(input("Input IPK Anda: "))

if ipk_3011 > 2.75:
    print("Anda lulus sangat memuaskan dengan IPK : " + str(ipk_3011))

    print("Program Selesai")
```

**Penjelasan**
Program ini mendemonstrasikan struktur if tanpa else. Perlu diperhatikan bahwa baris 'print("Program Selesai")' berada di dalam blok if (terindentasi), sehingga hanya tercetak jika kondisi terpenuhi. Ini berbeda dari program selanjutnya. Kekurangan program ini adalah tidak adanya penanganan untuk nilai IPK di bawah atau sama dengan 2.75.

### 2.2.2 Program "if2_2511533011"

```python
ipk_3011 = float(input("Input IPK Anda: "))

if ipk_3011 > 2.75:
    print("Anda lulus sangat memuaskan dengan IPK : " + str(ipk_3011))
else :
    print("Anda tidak lulus")

print("Program Selesai")
```

**Penjelasan :**
Struktur if-else memastikan salah satu dari dua blok kode selalu dieksekusi. Program ini lebih lengkap karena menangani kedua kemungkinan kondisi. 'Program Selesai' kini selalu tercetak karena berada di luar blok if-else, menunjukkan perbedaan penting dalam indentasi Python.

### 2.2.3 Program "if_elif_else1_2511533011"

```python
umur_3011 = int(input("Input umur anda : "))
sim_3011 = input("Apakah anda sudah memiliki sim C : ") [0]

if umur_3011 >= 17 and sim_3011 == 'y' :
    print("Anda sudah dewasa dan boleh bawa motor")

elif umur_3011 >= 17 and sim_3011 != 'y' :
    print("Anda suda dewasa tetapi tidak boleh membawa motor")

elif umur_3011 < 17 and sim_3011 == 'y' :
    print("Anda belum cukup umur tetapi sudah memiliki sim")

else : 
    print("Anda belum cukup umur dan tidak boleh membawa motor")
print("Program Selesai")
```

**Penjelasan :**
Struktur if-elif-else mengeksekusi tepat satu blok dari beberapa pilihan. Operator AND menggabungkan dua kondisi sekaligus. Penggunaan [0] pada input memastikan hanya karakter pertama yang diambil. Perbedaan utama dengan program berikutnya (multi_if1_2511533011) adalah bahwa elif-else bersifat eksklusif, sedangkan multiple if bersifat independen.

### 2.2.4 Program "multi_if1_2511533011"

```python
umur_3011 = int(input("input umur anda : "))
sim_3011 = input("Apakah anda sudah punya sim C (y/t) : ")[0]

if umur_3011 >= 17 and sim_3011 == 'y' :
    print("Anda sudah dewasa dan boleh bawa motor")

if umur_3011 >= 17 and sim_3011 != 'y' :
    print("Anda suda dewasa tetapi tidak boleh membawa motor")

if umur_3011 < 17 and sim_3011 == 'y' :
    print("Anda belum cukup umur tetapi sudah memiliki sim")

if umur_3011 < 17 and sim_3011 != 'y' :
    print("Anda belum cukup umur dan tidak boleh membawa motor")
```

**Penjelasan :**
Perbedaan kritis antara program ini dan program sebelumnya yaitu multiple if selalu mengevaluasi semua kondisi, sedangkan elif berhenti setelah menemukan kondisi pertama yang benar. Pada kasus ini hasil sama karena kondisi-kondisinya saling eksklusif. Namun jika ada tumpang tindih kondisi, multiple if bisa mengeksekusi lebih dari satu blok, sementara elif hanya satu.

### 2.2.5 Program "multi_if2_2511533011"

```python
total_belanja_3011 = float(input("Masukkan total belanja (Rp) : "))

input_member_3011 = input("Apakah anda memiliki member? (y/t) : ").strip().lower()
is_member_3011 = input_member_3011 in ["y", "ya"]

input_promo_3011 = input("Apakah kode promo anda valid? (y/t) : ").strip().lower()
kode_promo_valid_3011 = input_promo_3011 in ["y", "ya"]

total_diskon_persen_3011 = 0

if total_belanja_3011 > 1000000:
    total_diskon_persen_3011 += 10

if is_member_3011:
    total_diskon_persen_3011 += 5

if kode_promo_valid_3011 :
    total_diskon_persen_3011 += 15

nominal_diskon_3011 = total_belanja_3011 * (total_diskon_persen_3011 / 100)
total_bayar_3011 = total_belanja_3011 - nominal_diskon_3011
print ("\n-- Rincian Pembayaran --")
print (f"total Diskon : {total_diskon_persen_3011} %  (Rp {nominal_diskon_3011:,.0f})")
print (f"total Bayar : Rp {total_bayar_3011:,.0f}")

print (f"Total diskon yang anda dapatkan : {total_diskon_persen_3011}%")
```

**Penjelasan :**
Program ini menunjukkan penggunaan tepat multiple if independen karena memang memungkinkan lebih dari satu kondisi terpenuhi (diskon bersifat kumulatif). Operator += digunakan untuk akumulasi. Metode .strip().lower() memastikan input fleksibel terhadap spasi dan kapitalisasi. f-string dengan format :,.0f menghasilkan tampilan angka dengan pemisah ribuan.

### 2.2.6 Program "nilai_2511533011"

```python
nilai_3011 = int(input("Inputkan nilai angka = "))

if nilai_3011 >= 81 :
    print("A")
elif nilai_3011 >= 70 :
    print("B")
elif nilai_3011 >= 60 :
    print("C")
elif nilai_3011 >= 50 :
    print("D")

else :
    print('E')
```

**Penjelasan :**
Program ini memanfaatkan sifat if-elif secara efisien. Karena kondisi diperiksa secara berurutan dan berhenti di kondisi pertama yang benar, penulisan elif nilai >= 70 sudah cukup (implisit: nilai < 81) tanpa perlu menulis kondisi lengkap. Blok else menangani semua nilai di bawah 50 tanpa perlu menuliskan kondisi eksplisit.

### 2.2.7 Program "match_case1_2511533011"

```python
bulan_3011 = int(input("Masukkan angka bulan (1 - 12) : "))

match bulan_3011 :
    case 1 :
        print("Januari")
    case 2 :
        print("februari")
    case 3 :
        print('Maret')
    case 4 : 
        print("April")
    case 5 :
        print("Mei")
    case 6 :
        print("Juni")
    case 7 :
        print("Juli")
    case 8 :
        print('Agustus')
    case 9 :
        print("september")
    case 10 :
        print("Oktober")
    case 11 : 
        print("November")
    case 12 :
        print("Desember")
    case _:
        print('Angka Tidak Valid!')
```

**Penjelasan :**
Match-case adalah fitur baru Python 3.10+ yang lebih bersih dibanding if-elif panjang untuk pencocokan nilai tunggal. case _ adalah wildcard yang menangani semua nilai yang tidak cocok dengan case lain (setara dengan else). Match-case sangat berguna ketika banyak nilai diskrit harus ditangani.

# Tugas Praktikum

Soal :
Studi Kasus: Sistem Loket Terpadu & Audit Transaksi Ekspedisi Wahana
Sebuah wahana rekreasi petualangan membutuhkan sistem loket berbasis konsol untuk menentukan jenis wahana, memverifikasi kelaikan pengunjung dengan kondisi majemuk (logical operators), menghitung diskon akumulatif menggunakan multi-if terpisah, serta mencetak audit status operasional.
Catatan Materi: Seluruh alur program disusun murni menggunakan 4 pilar percabangan yang diajarkan pada modul Pekan 4: (1) if tunggal, (2) if-else, (3) if-elif-else dengan operator logika (and, or, !=), (4) Multi-if terpisah untuk diskon akumulasi, dan (5) match-case (1 s.d. 5 beserta case _). Tidak ada nested-if (if di dalam if).

**Kode Jawaban :**
a. Kode Program "tugas4_2511533011" 

```python
print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")

nama_3011       = input("Masukkan Nama Pengunjung: ")
umur_3011       = int(input("Input umur anda: "))
sim_3011        = input("Apakah Anda Sudah Punya SIM C (y/t): ").strip().lower()[0]
print()

print("Pilihan Paket Wahana (1-5):")
print("  1. Safari Rimba (Rp 50,000)")
print("  2. Arung Jeram (Rp 75,000)")
print("  3. Motor ATV Ekstrim (Rp 120,000)")
print("  4. Roller Coaster Kilat (Rp 100,000)")
print("  5. All-Access VIP (Rp 220,000)")

paket_3011       = int(input("Masukkan nomor paket (1-5): "))
jumlah_tiket_3011 = int(input("Masukkan jumlah tiket: "))
is_member_3011   = input("Apakah Anda member? (y/t): ").strip().lower()
kode_promo_3011  = input("Apakah kode promo valid? (y/t): ").strip().lower()

if jumlah_tiket_3011 <= 0:
    print("Peringatan: Kuota tiket tidak valid!")

nama_wahana_3011  = ""
harga_satuan_3011 = 0

match paket_3011:
    case 1:
        nama_wahana_3011  = "Wahana Safari Rimba"
        harga_satuan_3011 = 50000
    case 2:
        nama_wahana_3011  = "Wahana Arung Jeram"
        harga_satuan_3011 = 75000
    case 3:
        nama_wahana_3011  = "Wahana Motor ATV Ekstrim"
        harga_satuan_3011 = 120000
    case 4:
        nama_wahana_3011  = "Wahana Roller Coaster Kilat"
        harga_satuan_3011 = 100000
    case 5:
        nama_wahana_3011  = "Wahana All-Access VIP"
        harga_satuan_3011 = 220000
    case _:
        print("Paket wahana tidak valid!")

print()
print("--- KELAYAKAN PENGENDARA WAHANA ---")

if paket_3011 == 3:
    if umur_3011 >= 17 and sim_3011 == 'y':
        print("Status Akses: Anda sudah dewasa dan boleh mengendarai ATV sendiri.")
    elif umur_3011 >= 17 and sim_3011 != 'y':
        print("Status Akses: Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur).")
    elif umur_3011 < 17 and sim_3011 == 'y':
        print("Status Akses: Identitas tidak valid: Belum cukup umur memiliki SIM.")
    else:
        print("Status Akses: Anda belum cukup umur dan tidak boleh bawa motor ATV.")
else:
    # if-else sederhana untuk paket selain 3
    if umur_3011 >= 10:
        print("Status Akses: Anda memenuhi syarat usia untuk wahana ini.")
    else:
        print("Status Akses: Anda belum memenuhi syarat usia minimum (10 tahun) untuk wahana ini.")

subtotal_3011          = harga_satuan_3011 * jumlah_tiket_3011
total_diskon_persen_3011 = 0

if subtotal_3011 >= 200000:
    total_diskon_persen_3011 += 10

if is_member_3011 in ['y', 'ya']:
    total_diskon_persen_3011 += 5

if kode_promo_3011 in ['y', 'ya']:
    total_diskon_persen_3011 += 15

if jumlah_tiket_3011 >= 5:
    total_diskon_persen_3011 += 5

nominal_diskon_3011 = subtotal_3011 * (total_diskon_persen_3011 / 100)
total_bayar_3011    = subtotal_3011 - nominal_diskon_3011

if total_bayar_3011 > 300000:
    catatan_3011 = "Selamat! Anda berhak mendapatkan Souvenir Gratis."
else:
    catatan_3011 = "Terima kasih telah berkunjung."

print()
print("--- Rincian Pembayaran ---")
print(f"Subtotal Belanja: Rp {subtotal_3011:,.0f}")
print(f"Total Diskon: {total_diskon_persen_3011}% (Rp {nominal_diskon_3011:,.0f})")
print(f"Total Bayar: Rp {total_bayar_3011:,.0f}")
print(f"Catatan Layanan: {catatan_3011}")
print("Program Selesai")
```

**Penjelasan :**

Program Sistem Loket Alpro Adventure Park adalah program berbasis konsol yang mensimulasikan sistem pembelian tiket wahana rekreasi. Program dimulai dengan meminta input data pengunjung berupa nama, umur, kepemilikan SIM C, pilihan paket wahana (1–5), jumlah tiket, status member, dan kode promo.

Setelah input diterima, program memvalidasi jumlah tiket menggunakan if tunggal. Jika tiket yang dimasukkan kurang dari atau sama dengan nol, program mencetak peringatan. Selanjutnya, program menentukan nama wahana dan harga satuan menggunakan match-case berdasarkan nomor paket yang dipilih, jika nomor di luar 1–5, program mencetak pesan tidak valid.

Program kemudian memverifikasi kelayakan pengunjung. Khusus paket 3 (Motor ATV Ekstrim), verifikasi dilakukan menggunakan if-elif-else dengan operator logika and dan != untuk mengecek kombinasi umur dan kepemilikan SIM. Untuk paket lainnya, cukup dicek apakah umur pengunjung memenuhi batas minimum 10 tahun menggunakan if-else sederhana.

Diskon dihitung menggunakan empat blok if independen yang berjalan secara terpisah sehingga diskon dapat terakumulasi: diskon belanja besar (10%), diskon member (5%), diskon voucher promo (15%), dan diskon rombongan (5%), dengan total diskon maksimal 35%. Terakhir, program menghitung total bayar dan menggunakan if-else untuk menentukan apakah pengunjung berhak mendapat souvenir gratis, kemudian mencetak seluruh rincian pembayaran menggunakan f-string.
