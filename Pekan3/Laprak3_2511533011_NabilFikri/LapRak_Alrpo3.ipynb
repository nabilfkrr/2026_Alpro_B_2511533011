## 2.1 Dasar Teori
Operator adalah simbol khusus dalam bahasa pemrograman yang digunakan untuk melakukan suatu operasi terhadap satu atau lebih operand (nilai/variabel). Python menyediakan beberapa kelompok operator, di antaranya :

| No | Kelompok Operator | Contoh Simbol | Fungsi |
|----|--------------------|----------------|--------|
| 1  | Aritmatika | `+ - * / // % **` | Melakukan operasi hitung matematika dasar |
| 2  | Perbandingan (Relasional) | `> < >= <= == !=` | Membandingkan dua nilai, menghasilkan `bool` |
| 3  | Assignment | `= += -= *= /= //= %= **=` | Memberikan atau memperbarui nilai suatu variabel |
| 4  | Logika (Logical) | `and or not` | Menggabungkan/membalik nilai boolean |
| 5  | Bitwise | `& \| ^ ~ << >>` | Beroperasi langsung pada representasi biner bilangan |
| 6  | Keanggotaan (Membership) | `in`, `not in` | Memeriksa apakah suatu nilai ada dalam suatu koleksi (list, tuple, dsb.) |
| 7  | Identitas (Identity) | `is`, `is not` | Memeriksa apakah dua variabel merujuk pada objek yang sama di memori |

Pemahaman terhadap operator ini penting karena menjadi dasar dalam membangun ekspresi logika dan alur kendali (percabangan maupun perulangan) pada program yang lebih kompleks.

## 2.2 Kode Program

### 2.2.1 Program "aritmatika_2511533011"

```python
angka1_3011 = int(input("input angka -1 : "))
angka2_3011 = int(input("input angka -2 : "))

# Penjumlahan
hasil_3011 = angka1_3011 + angka2_3011
print("\nOperator Penjumlahan")
print("Hasil =", hasil_3011)

# Pengurangan
hasil_3011 = angka1_3011 - angka2_3011
print("\nOperator Pengurangan")
print("Hasil =", hasil_3011)

# Perkalian
hasil_3011 = angka1_3011 * angka2_3011
print("\nOperator Perkalian")
print("Hasil =", hasil_3011)

if angka2_3011 != 0:

    # Pembagian
    hasil_3011 = angka1_3011 / angka2_3011
    print("\nOperator Pembagian")
    print("Hasil =", hasil_3011)

    # Modulus / Sisa Bagi    
    hasil_3011 = angka1_3011 % angka2_3011
    print("\nOperator Modulus")
    print("Hasil =", hasil_3011)


    # Pembagian bilangan bulat    
    hasil_3011 = angka1_3011 // angka2_3011
    print("\nOperator Pembagian Bulat")
    print("Hasil =", hasil_3011)

else :
    print("Angka kedua tidal boleh bernilai 0.")
    
 # Pangkat
hasil_3011 = angka1_3011 ** angka2_3011
print("\nOperator Pangkat")
print("Hasil =", hasil_3011)
```

**Penjelasan :**
Program berikut menghitung hasil dari operasi penjumlahan (`+`), pengurangan (`-`), perkalian (`*`), pembagian (`/`), modulus/sisa bagi (`%`), pembagian bulat (`//`), dan pemangkatan (`**`) dari dua bilangan bulat yang diinput pengguna. Program juga menyertakan pengecekan `if angka2 != 0` untuk mencegah kesalahan input nilai 0 pada operasi pembagian.

Pada hasil pembagian (`/`) menghasilkan tipe data `float` (3.4), sedangkan pembagian bulat (`//`) membulatkan hasil ke bawah menjadi `3`. Operator modulus (`%`) mengembalikan sisa hasil bagi (`2`), dan operator pangkat (`**`) menghasilkan nilai yang jauh lebih besar karena bersifat eksponensial (17⁵ = 1.419.857).

### 2.2.2 Program "perbandingan_2511533011"

```python
angka1_3011= int(input("input angka -1 : "))
angka2_3011= int(input("input angka -2 : "))

hasil_3011 = angka1_3011 > angka2_3011
print("\nOperator lebih besar dari")
print("angka1 > angka2 =", hasil_3011)

hasil_3011 = angka1_3011 < angka2_3011
print("\nOperator lebih kecil dari")
print("angka1 < angka2 =", hasil_3011)

hasil_3011 = angka1_3011 >= angka2_3011
print("\nOperator lebih besar dari atau sama dengan")
print("angka1 >= angka2 =", hasil_3011)

hasil_3011 = angka1_3011 <= angka2_3011
print("\nOperator lebih kecil dari atau sama dengan")
print("angka1 <= angka2 =", hasil_3011)

hasil_3011 = angka1_3011 == angka2_3011
print("\nOperator sama dengan")
print("angka1 == angka2 =", hasil_3011)

hasil_3011 = angka1_3011 != angka2_3011
print("\nOperator tidak sama dengan")
print("angka1 != angka2 =", hasil_3011)

hasil_3011 = 0 < angka1_3011 < 100
print("\nPerbandingan berantai")
print("0 < angka1 < 100 =", hasil_3011)

hasil_3011 = 0 < angka2_3011 < 100
print("0 < angka2 < 100 =", hasil_3011)
```

**Penjelasan :**
Program ini membandingkan dua bilangan menggunakan operator relasional (`>`, `<`, `>=`, `<=`, `==`, `!=`) serta menunjukkan konsep *perbandingan berantai* (`0 < angka < 100`) yang merupakan salah satu keunikan sintaks Python.

Dilihat pada hasil (output) di ata, setiap operasi perbandingan menghasilkan nilai bertipe `bool` (`True`/`False`). Perbandingan berantai `0 < angka1 < 100` secara internal setara dengan `(0 < angka1) and (angka1 < 100)`, sehingga lebih ringkas dibanding menuliskannya secara terpisah.

### 2.2.3 Program "assignment_2511533011"

```python
angka1_3011 = int(input("Input angka-1: "))
angka2_3011 = int(input("Input angka-2: "))

print("\nNilai awal angka1 =", angka1_3011)
print("Nilai angka2 =", angka2_3011)

# Assignment biasa
hasil_3011 = angka1_3011
print("\nAssignment biasa (=)")
print("Hasil =", hasil_3011)

# Assignment penambahan
hasil_3011 = angka1_3011
hasil_3011 += angka2_3011
print("\nAssignment penambahan (+=)")
print("Hasil =", hasil_3011)

# Assignment pengurangan
hasil_3011 = angka1_3011
hasil_3011 -= angka2_3011
print("\nAssignment pengurangan (-=)")
print("Hasil =", hasil_3011)

# Assignment perkalian
hasil_3011 = angka1_3011
hasil_3011 *= angka2_3011
print("\nAssignment perkalian (*=)")
print("Hasil =", hasil_3011)

# Assignment pembagian, pembagian bulat, dan sisa bagi
if angka2_3011 != 0:
    hasil_3011 = angka1_3011
    hasil_3011 /= angka2_3011
    print("\nAssignment pembagian (/=)")
    print("Hasil =", hasil_3011)
    
    # Operator tambahan
    hasil_3011 = angka1_3011
    hasil_3011 //= angka2_3011
    print("\nAssignment pembagian bulat (//=)")
    print("Hasil =", hasil_3011)
    
    hasil_3011 = angka1_3011
    hasil_3011 %= angka2_3011
    print("\nAssignment sisa bagi (%=)")
    print("Hasil =", hasil_3011)
else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka kedua tidak boleh bernilai 0.")

# Operator tambahan: assignment perpangkatan
hasil_3011 = angka1_3011
hasil_3011 **= angka2_3011
print("\nAssignment perpangkatan (**=)")
print("Hasil =", hasil_3011)
```

**Penjelasan :**
Program ini mendemonstrasikan operator assignment gabungan (*compound assignment*) seperti `+=`, `-=`, `*=`, `/=`, `//=`, `%=`, dan `**=`, yang merupakan bentuk singkat dari `variabel = variabel operator nilai`.

Setiap variasi assignment gabungan menghasilkan nilai akhir yang sama dengan versi operator biasa yang setara (misalnya `hasil += angka2` sama dengan `hasil = hasil + angka2`), namun penulisannya lebih ringkas dan umum digunakan dalam praktik pemrograman sehari-hari.  

### 2.2.4 Program "lainnya_2511533011"

```python
print("=======================================")
print("1. OPERATOR KEANGGOTAAN")
print("=======================================")

# Input beberapa data yang dipisahkan dengan koma
input_data_3011 = input("Masukkan beberapa angka, pisahkan dengan koma: ")

# Mengubah input menjadi list integer
data_3011 = [int(angka_3011.strip()) for angka_3011 in input_data_3011.split(",")]

nilai_dicari_3011 = int(input("Masukkan angka yang ingin dicari: "))

# Operator in
hasil_3011 = nilai_dicari_3011 in data_3011
print("\nOperator keanggotaan IN")
print(nilai_dicari_3011, "in", data_3011, "=", hasil_3011)

# Operator not in
hasil_3011 = nilai_dicari_3011 not in data_3011
print("\nOperator keanggotaan NOT IN")
print(nilai_dicari_3011, "not in", data_3011, "=", hasil_3011)

print("\n=======================================")
print("2. OPERATOR IDENTITAS")
print("=======================================")

# objek1 menggunakan list dari input pengguna
objek1_3011 = data_3011

# objek2 merujuk pada objek yang sama dengan objek1
objek2_3011 = objek1_3011

# objek3 memiliki isi sama, tetapi merupakan objek baru
objek3_3011 = data_3011.copy()

print("objek1 =", objek1_3011)
print("objek2 =", objek2_3011)
print("objek3 =", objek3_3011)

# Operator is
hasil_3011 = objek1_3011 is objek2_3011
print("\nOperator identitas IS")
print("objek1 is objek2 =", hasil_3011)

# Operator is not
hasil_3011 = objek1_3011 is not objek3_3011
print("\nOperator identitas IS NOT")
print("objek1 is not objek3 =", hasil_3011)

# Membandingkan identitas dan nilai
print("\nPerbandingan identitas dan nilai")
print("objek1 is objek3 =", objek1_3011 is objek3_3011)
print("objek1 == objek3 =", objek1_3011 == objek3_3011)
```

**Penjelasan :**
Program ini dibagi menjadi dua bagian. Bagian pertama menggunakan operator keanggotaan `in` dan `not in` untuk memeriksa apakah suatu nilai terdapat dalam sebuah list. Bagian kedua menggunakan operator identitas `is` dan `is not` untuk membandingkan apakah dua variabel menunjuk pada objek yang sama di memori, dibandingkan dengan operator kesamaan nilai `==`.

pada program ini, `objek2 = objek1` membuat `objek2` menunjuk ke alamat memori yang sama dengan `objek1`, sehingga `objek1 is objek2` bernilai `True`. Sebaliknya, `objek3 = objek1.copy()` membuat objek baru dengan isi identik namun berbeda alamat memori, sehingga `objek1 is objek3` bernilai `False`, meskipun secara nilai (`==`) keduanya tetap `True`. Ini menegaskan perbedaan mendasar antara *kesamaan nilai* (`==`) dan *kesamaan identitas* (`is`).

### 2.2.5 Program "logika_2511533011"

```python
a1_3011 = input("Input nilai boolean-1 (true/false): ").strip().lower() == "true"
a2_3011 = input("Input nilai boolean-2 (true/false): ").strip().lower() == "true"

print("\nA1 =", a1_3011)
print("A2 =", a2_3011)

# Konjungsi: bernilai True jika keduanya True
hasil_3011 = a1_3011 and a2_3011
print("\nKonjungsi (AND)")
print("A1 and A2 =", hasil_3011)

# Disjungsi: bernilai True jika salah satunya True
hasil_3011 = a1_3011 or a2_3011
print("\nDisjungsi (OR)")
print("A1 or A2 =", hasil_3011)

# Negasi A1: membalik nilai A1
hasil_3011 = not a1_3011
print("\nNegasi A1 (NOT)")
print("not A1 =", hasil_3011)

# Negasi A2: membalik nilai A2
hasil_3011 = not a2_3011
print("\nNegasi A2 (NOT)")
print("not A2 =", hasil_3011)

# XOR: bernilai True jika kedua nilai berbeda
hasil_3011 = a1_3011 != a2_3011
print("\nDisjungsi Eksklusif (XOR)")
print("A1 XOR A2 =", hasil_3011)
```

### 2.2.6 Program "bitwise_2511533011"

```python
print("\n=======================================")
print("3. OPERATOR BITWISE")
print("=======================================")

angka1_3011 = int(input("Masukkan angka bitwise-1: "))
angka2_3011 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1 =", angka1_3011, "| biner =", bin(angka1_3011))
print("angka2 =", angka2_3011, "| biner =", bin(angka2_3011))

# Bitwise AND
hasil_3011 = angka1_3011 & angka2_3011
print("\nBitwise AND (&)")
print(angka1_3011, "&", angka2_3011, "=", hasil_3011)
print("Biner hasil =", bin(hasil_3011))
print("Biner hasil (8 bit) =", format(hasil_3011, "08b"))

# Bitwise OR
hasil_3011 = angka1_3011 | angka2_3011
print("\nBitwise OR (|)")
print(angka1_3011, "|", angka2_3011, "=", hasil_3011)
print("Biner hasil =", bin(hasil_3011))
print("Biner hasil (8 bit) =", format(hasil_3011, "08b"))

# Bitwise XOR
hasil_3011 = angka1_3011 ^ angka2_3011
print("\nBitwise XOR (^)")
print(angka1_3011, "^", angka2_3011, "=", hasil_3011)
print("Biner hasil =", bin(hasil_3011))
print("Biner hasil (8 bit) =", format(hasil_3011, "08b"))

# Bitwise NOT
hasil_3011 = ~angka1_3011
print("\nBitwise NOT (~)")
print("~", angka1_3011, "=", hasil_3011)
print("Biner hasil =", bin(hasil_3011))
print("Biner hasil (8 bit) =", format(hasil_3011, "08b"))

# Bitwise geser kiri
jumlah_geser_3011 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_3011 = angka1_3011 << jumlah_geser_3011
print("\nBitwise geser kiri (<<)")
print(angka1_3011, "<<", jumlah_geser_3011, "=", hasil_3011)
print("Biner hasil =", bin(hasil_3011))
print("Biner hasil (8 bit) =", format(hasil_3011, "08b"))

# Bitwise geser kanan
hasil_3011 = angka1_3011 >> jumlah_geser_3011
print("\nBitwise geser kanan (>>)")
print(angka1_3011, ">>", jumlah_geser_3011, "=", hasil_3011)
print("Biner hasil =", bin(hasil_3011))
print("Biner hasil (8 bit) =", format(hasil_3011, "08b"))
```

**Penjelasan :**
Program ini menerima dua input nilai boolean, kemudian menerapkan operator logika `and` (konjungsi), `or` (disjungsi), `not` (negasi), serta operator `!=` yang digunakan untuk mensimulasikan XOR (*exclusive or*).

Operator `and` hanya bernilai `True` jika kedua operand bernilai `True`, sedangkan `or` bernilai `True` jika salah satu operand `True`. Karena Python tidak memiliki operator XOR bawaan untuk boolean, operator `!=` dimanfaatkan sebagai penggantinya karena XOR bernilai `True` hanya ketika kedua nilai berbeda.

```python
print("=== SISTEM TRANSAKSI TOKO ===")

# INPUT
nama_3011           = input("\nMasukkan Nama Pelanggan : ")
status_3011         = input("Masukkan Status Pelanggan (member/nonmember) : ")
totalBelanja_3011   = int(input("Masukkan Total Belanja : "))
jumlahBarang_3011   = int(input("Masukkan Jumlah Barang : "))
kodePromo_3011      = input("Masukkan Kode Promo : ").strip().upper()

# KONSTANTA
minBelanja_3011     = 200000
minBarang_3011      = 3
diskonMember_3011   = 10

daftarPromo_3011    = ["MERIAH10", "MERIAH20", "GRATISONGKIR"]
diskonPromo_3011    = {"MERIAH10": 10, "MERIAH20": 20, "GRATISONGKIR": 0}

# OPERATOR PERBANDINGAN
syaratBelanja_3011  = totalBelanja_3011 >= minBelanja_3011
syaratBarang_3011   = jumlahBarang_3011 >= minBarang_3011
isMember_3011       = status_3011.lower() == "member"

# OPERATOR KEANGGOTAAN
promoTersedia_3011  = kodePromo_3011 in daftarPromo_3011
promoTidakAda_3011  = kodePromo_3011 not in daftarPromo_3011

# OPERATOR LOGIKA
dapatDiskon_3011    = isMember_3011 and syaratBelanja_3011
dapatPromo_3011     = promoTersedia_3011 and (syaratBelanja_3011 or syaratBarang_3011)
cekNot_3011         = not promoTidakAda_3011

# OPERATOR ARITMATIKA
persenDiskon_3011   = 0
if dapatPromo_3011:
    persenDiskon_3011 = diskonPromo_3011[kodePromo_3011]
elif dapatDiskon_3011:
    persenDiskon_3011 = diskonMember_3011

besarDiskon_3011    = totalBelanja_3011 * persenDiskon_3011 // 100
totalBayar_3011     = totalBelanja_3011 - besarDiskon_3011
rataRata_3011       = totalBelanja_3011 // jumlahBarang_3011
sisaBelanja_3011    = totalBelanja_3011 % jumlahBarang_3011
point_3011          = totalBelanja_3011 // 10000

# OPERATOR PENUGASAN
ongkir_3011         = 15000
if kodePromo_3011 == "GRATISONGKIR" and promoTersedia_3011:
    ongkir_3011     -= ongkir_3011     
totalBayar_3011     += ongkir_3011     

# OPERATOR IDENTITAS
refStatus_3011      = status_3011           
kopiStatus_3011     = status_3011[:]       
identitasSama_3011  = refStatus_3011 is status_3011
identitasBeda_3011  = kopiStatus_3011 is not status_3011
nilaiSama_3011      = kopiStatus_3011 == status_3011

# OPERATOR BITWISE

bitMember_3011      = 0b0001 if isMember_3011      else 0b0000
bitBelanja_3011     = 0b0010 if syaratBelanja_3011 else 0b0000
bitBarang_3011      = 0b0100 if syaratBarang_3011  else 0b0000
bitPromo_3011       = 0b1000 if promoTersedia_3011 else 0b0000

kodeTranasksi_3011  = bitMember_3011 | bitBelanja_3011 | bitBarang_3011 | bitPromo_3011  
cekMemberBit_3011   = kodeTranasksi_3011 & 0b0001  
cekPromoBit_3011    = kodeTranasksi_3011 & 0b1000  
kodeReferensi_3011  = 0b1011
bedaStatus_3011     = kodeTranasksi_3011 ^ kodeReferensi_3011   
shiftKiri_3011      = kodeTranasksi_3011 << 1                   

akseMember_3011     = bool(kodeTranasksi_3011 & 0b0001)
aksePromo_3011      = bool(kodeTranasksi_3011 & 0b1000)
akseFreeShip_3011   = kodePromo_3011 == "GRATISONGKIR" and promoTersedia_3011

print("\n=== DATA TRANSAKSI ===")
print(f"Nama Pelanggan   : {nama_3011}")
print(f"Status Pelanggan : {status_3011}")
print(f"Total Belanja    : Rp{totalBelanja_3011:,}")
print(f"Jumlah Barang    : {jumlahBarang_3011}")
print(f"Kode Promo       : {kodePromo_3011}")

 
print("\n=== HASIL VALIDASI ===")
  
print(f"Belanja >= Rp200.000       : {syaratBelanja_3011}")
print(f"Jumlah Barang >= 3         : {syaratBarang_3011}")
print(f"Status Member              : {isMember_3011}")
print(f"Kode Promo Tersedia        : {promoTersedia_3011}")
print(f"Mendapatkan Diskon Member  : {dapatDiskon_3011}")
print(f"Mendapatkan Promo          : {dapatPromo_3011}")

 
print("\n=== HASIL PERHITUNGAN ===")
  
print(f"Persentase Diskon          : {persenDiskon_3011}%")
print(f"Besarnya Diskon            : Rp{besarDiskon_3011:,}")
print(f"Ongkos Kirim               : Rp{ongkir_3011:,}")
print(f"Total Pembayaran           : Rp{totalBayar_3011:,}")
print(f"Rata-rata Harga Barang     : Rp{rataRata_3011:,}")
print(f"Sisa Pembagian (%)         : Rp{sisaBelanja_3011:,}")
print(f"Poin Pelanggan             : {point_3011} poin")

 
print("\n=== HAK AKSES PELANGGAN ===")
  
print(f"Kode Hak Akses (desimal)   : {kodeTranasksi_3011}")
print(f"Kode Hak Akses (biner)     : {format(kodeTranasksi_3011, '04b')}")
print(f"Member Access              : {akseMember_3011}")
print(f"Promo Access               : {aksePromo_3011}")
print(f"Free Shipping Access       : {akseFreeShip_3011}")

 
print("\n=== OPERATOR IDENTITAS ===")
  
print(f"refStatus is status_3011       : {identitasSama_3011}")
print(f"kopiStatus is not status_3011  : {identitasBeda_3011}")
print(f"kopiStatus == status_3011      : {nilaiSama_3011}")
print("-> is  : membandingkan IDENTITAS objek di memori")
print("-> ==  : membandingkan NILAI/ISI objek")

 
print("\n=== OPERASI BITWISE ===")
  
print("\n=== Kode Status Transaksi ===")
print(f"  0001 | 0010 | 0100 | 1000")
print(f"  Kode Biner   : {format(kodeTranasksi_3011, '04b')}")
print(f"  Kode Desimal : {kodeTranasksi_3011}")

print("\n=== Pemeriksaan Status ===")
print(f"  Cek Member : {format(kodeTranasksi_3011,'04b')} & 0001 = {format(cekMemberBit_3011,'04b')} ({cekMemberBit_3011})")
print(f"  Cek Promo  : {format(kodeTranasksi_3011,'04b')} & 1000 = {format(cekPromoBit_3011,'04b')} ({cekPromoBit_3011})")

print("\n=== Perbandingan Status ===")
print(f"  Kode Transaksi : {format(kodeTranasksi_3011,'04b')}")
print(f"  Kode Referensi : {format(kodeReferensi_3011,'04b')}")
print(f"  Hasil      : {format(bedaStatus_3011,'04b')} ({bedaStatus_3011})")

print("\n=== Shift ===")
print(f"  {format(kodeTranasksi_3011,'04b')} << 1")
print(f"  Hasil Biner   : {bin(shiftKiri_3011)[2:]}")
print(f"  Hasil Desimal : {shiftKiri_3011}")

 
print("=== SELESAI ===")
```

**Penjelasan :**

Pada program ini, demonstrasi operator identitas menggunakan dua variabel turunan dari `status_3011`, yaitu `refStatus_3011` dan `kopiStatus_3011`. Variabel `refStatus_3011` dibuat dengan cara `refStatus_3011 = status_3011`, yang artinya kedua variabel menunjuk ke **objek yang sama persis** di memori, sehingga `refStatus_3011 is status_3011` menghasilkan `True`.

Sementara itu, `kopiStatus_3011` dibuat dengan cara `status_3011[:]` yang seharusnya menghasilkan salinan objek baru. Namun karena Python memiliki mekanisme bernama **string interning**, yaitu optimasi di mana Python menyimpan string pendek yang sering dipakai di satu lokasi memori yang sama, maka salinan string tersebut bisa saja tetap menunjuk ke objek yang sama dengan aslinya. Akibatnya, `kopiStatus_3011 is not status_3011` bisa menghasilkan `False`, bukan `True` seperti yang diharapkan.

Perlu ditekankan bahwa ini bukan bug dalam program, melainkan perilaku internal Python yang bekerja di balik layar. Nilai dari kedua variabel tetap sama (`==` menghasilkan `True`), dan perbedaan antara `is` dengan `==` tetap dapat dijelaskan dengan benar, yaitu bahwa `is` membandingkan identitas objek di memori, sedangkan `==` membandingkan nilai atau isi dari objek tersebut. Jika ingin hasil `is not` yang selalu konsisten `True`, disarankan menggunakan tipe data `list` sebagai objek yang dibandingkan, karena `list` tidak mengalami string interning.

# F. Pertanyaan Analisis pada Laporan

1. Operator aritmatika yang digunakan dan fungsinya dalam menghitung transaksi

Program ini menggunakan enam operator aritmatika secara terintegrasi. Operator perkalian (*) digunakan untuk menghitung nilai diskon dengan mengalikan total belanja dengan persentase diskon. Operator pembagian bulat (//) digunakan dua kali, yaitu untuk menghasilkan nilai diskon dalam bentuk bilangan bulat (dibagi 100) dan untuk menghitung rata-rata harga barang. Operator pengurangan (-) digunakan untuk mendapatkan total bayar setelah diskon dikurangkan dari total belanja. Operator modulus (%) digunakan untuk menghitung sisa pembagian total belanja dengan jumlah barang. Operator penjumlahan (+) digunakan melalui augmented assignment += untuk menambahkan ongkos kirim ke total pembayaran. Terakhir, operator // juga digunakan untuk menghitung poin pelanggan, di mana setiap kelipatan Rp10.000 menghasilkan satu poin.

2. Perbedaan operator and, or, dan not berdasarkan penerapannya

Operator and digunakan pada kondisi dapatDiskon_3011 = isMember_3011 and syaratBelanja_3011, yang berarti pelanggan hanya mendapatkan diskon member apabila kedua kondisi terpenuhi sekaligus, yaitu berstatus member dan total belanjanya mencapai batas minimum. Operator or digunakan pada kondisi syaratBelanja_3011 or syaratBarang_3011, yang berarti cukup salah satu syarat saja yang terpenuhi agar pelanggan memenuhi syarat promo. Operator not digunakan pada cekNot_3011 = not promoTidakAda_3011, yang membalik nilai boolean promoTidakAda_3011 — jika kode promo tidak ada maka nilainya True, setelah dibalik dengan not menjadi False, yang artinya promo tidak tersedia.

3. Fungsi operator in dan not in pada pemeriksaan kode promo

Operator in digunakan untuk memeriksa apakah kode promo yang dimasukkan pelanggan terdapat di dalam daftarPromo_3011. Jika kode promo ditemukan, variabel promoTersedia_3011 bernilai True dan program akan memproses diskon sesuai kode tersebut. Sebaliknya, operator not in digunakan untuk memeriksa kondisi kebalikannya, yaitu apabila kode promo tidak ditemukan dalam daftar, maka promoTidakAda_3011 bernilai True. Penggunaan kedua operator ini membuat pemeriksaan kode promo menjadi sangat ringkas tanpa perlu membandingkan satu per satu menggunakan rangkaian if-elif.

4. Perbedaan antara is dan == pada Python

Operator == membandingkan nilai atau isi dari dua objek, sedangkan operator is membandingkan identitas objek, yaitu apakah dua variabel menunjuk ke lokasi memori yang sama. Dalam program ini, refStatus_3011 = status_3011 membuat kedua variabel menunjuk ke objek yang sama di memori, sehingga refStatus_3011 is status_3011 menghasilkan True. Sementara kopiStatus_3011 dibuat sebagai salinan menggunakan slicing [:], sehingga secara konsep merupakan objek berbeda namun memiliki nilai yang sama, maka kopiStatus_3011 == status_3011 menghasilkan True karena nilainya identik. Perlu dicatat bahwa pada string pendek, Python dapat melakukan string interning sehingga hasil is bisa tetap True meskipun dibuat sebagai salinan, namun hal ini merupakan perilaku internal Python dan bukan kesalahan program.

5. Operator bitwise untuk menentukan hak akses pelanggan

Setiap kondisi pelanggan direpresentasikan sebagai satu bit dalam bilangan biner 4-bit, di mana bit ke-0 (0001) mewakili status member, bit ke-1 (0010) mewakili syarat belanja, bit ke-2 (0100) mewakili syarat jumlah barang, dan bit ke-3 (1000) mewakili ketersediaan kode promo. Keempat bit tersebut kemudian digabungkan menggunakan operator OR (|) menjadi satu kode transaksi utuh. Untuk memeriksa hak akses tertentu, digunakan operator AND (&) — misalnya kodeTransaksi_3011 & 0b0001 akan menghasilkan nilai lebih dari nol hanya jika bit member aktif. Operator XOR (^) digunakan untuk membandingkan kode transaksi dengan kode referensi guna mengetahui bit mana yang berbeda. Terakhir, operator shift kiri (<<) menggeser seluruh bit satu posisi ke kiri, yang secara matematis setara dengan mengalikan nilai dengan 2.

6. Alur program dari input hingga total pembayaran dan status promo

Program dimulai dengan meminta pengguna memasukkan lima data, yaitu nama, status pelanggan, total belanja, jumlah barang, dan kode promo. Setelah input diterima, program melakukan serangkaian pengecekan menggunakan operator perbandingan untuk menentukan apakah syarat belanja dan syarat jumlah barang terpenuhi, serta apakah pelanggan berstatus member. Kemudian operator keanggotaan digunakan untuk memverifikasi apakah kode promo yang dimasukkan ada dalam daftar promo yang berlaku. Hasil pengecekan tersebut kemudian digabungkan menggunakan operator logika untuk menentukan apakah pelanggan berhak mendapat diskon member, promo kode, atau tidak mendapat diskon sama sekali. Berdasarkan hasil logika tersebut, program menghitung besarnya diskon, menguranginya dari total belanja, lalu menambahkan ongkos kirim menggunakan operator augmented assignment. Hasil akhirnya adalah total pembayaran yang sudah memperhitungkan diskon dan ongkos kirim.

7. Kondisi yang menghasilkan True dan False

Dengan contoh input nama Anita, status member, total belanja Rp250.000, jumlah barang 4, dan kode promo MERIAH10 — kondisi syaratBelanja_3011 = totalBelanja_3011 >= minBelanja_3011 menghasilkan True karena Rp250.000 memang lebih besar dari batas minimum Rp200.000, sehingga pelanggan memenuhi syarat belanja. Sebaliknya, kondisi akseFreeShip_3011 = kodePromo_3011 == "GRATISONGKIR" and promoTersedia_3011 menghasilkan False karena kode promo yang digunakan adalah MERIAH10, bukan GRATISONGKIR, sehingga syarat pertama dari kondisi and sudah gugur dan pelanggan tidak mendapatkan gratis ongkir.
