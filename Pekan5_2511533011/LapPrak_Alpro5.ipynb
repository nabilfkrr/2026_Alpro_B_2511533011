## 2.1 Dasar Teori Perulangan for
Perulangan for pada Python menelusuri setiap elemen dari sebuah barisan nilai (iterable) dan menjalankan blok kode di dalamnya satu kali untuk setiap elemen. Blok kode ditandai dengan indentasi.

### 2.1.1 Fungsi range()
range() menghasilkan barisan bilangan bulat. Bentuk range(n) menghasilkan 0 sampai n - 1, sedangkan range(awal, akhir) menghasilkan bilangan dari awal sampai akhir - 1. Batas akhir bersifat eksklusif, sehingga untuk memperoleh 1 sampai n digunakan range(1, n + 1).

### 2.1.2 Parameter end pada print()
Secara bawaan print() mengakhiri keluaran dengan pindah baris. Parameter end mengganti karakter penutup tersebut, misalnya end = " " agar keluaran berikutnya tercetak pada baris yang sama.

### 2.1.3 Akumulator dan Operator Modulus
Akumulator adalah variabel yang nilainya ditambah terus pada tiap iterasi (misalnya jumlah += i) untuk menghimpun total. Operator modulus (%) menghasilkan sisa bagi; ekspresi i % 2 == 0 bernilai benar untuk bilangan genap.

### 2.1.4 Perulangan Bersarang (Nested Loop)
Perulangan bersarang adalah perulangan yang berada di dalam perulangan lain. Perulangan dalam dijalankan sepenuhnya untuk setiap satu iterasi perulangan luar, dan umum dipakai untuk membentuk pola baris dan kolom.


## 2.2 Kode Program

### 2.2.1 Program "perulangan_for1_2511533011"

```python
ulang_3011 = int(input("Masukkan jumlah perulangan: "))

for i__3011 in range(ulang_3011):
    print(f"Perulangan ke-{i__3011}")
```

**Penjelasan**
Program meminta pengguna memasukkan jumlah perulangan, lalu menyimpannya sebagai bilangan bulat pada variabel ulang_3011 lewat int(input(...)).
Perulangan for dengan range(ulang_3011) menghasilkan nilai counter mulai dari 0 sampai ulang_3011 - 1, sehingga jumlah iterasi tepat sama dengan nilai yang dimasukkan. Pada setiap iterasi, f-string mencetak "Perulangan ke-" diikuti nilai counter.
Karena range() dimulai dari 0, perulangan pertama diberi nomor 0 dan perulangan terakhir bernomor n - 1.


### 2.2.2 Program "perulangan_for2_2511533011"

```python
ulang_3011 = int(input("Masukkan jumlah perulangan: "))
print("Perulangan ke-0 sampai ke", ulang_3011 - 1)
for i__3011 in range(ulang_3011):
    print(i__3011, end = " ")
print()
print("Perulangan ke-1 sampai ke-", ulang_3011)
for i__3011 in range(1, ulang_3011 + 1):
    print(i__3011, end = " ")
```

**Penjelasan :**
Program membandingkan dua cara memakai range(). Cara pertama, range(n), menghasilkan 0 sampai n - 1. Cara kedua, range(1, n + 1), menghasilkan 1 sampai n karena batas atas pada range() bersifat eksklusif sehingga perlu ditambah 1.
Parameter end = " " pada print() mengganti karakter akhir baris bawaan (newline) dengan spasi, sehingga seluruh angka tercetak dalam satu baris. print() kosong di antara dua perulangan berfungsi pindah baris.


### 2.2.3 Program "perulangan_for3_2511533011"

```python
ulang_3011 = int(input("Masukkan jumlah perulangan: "))

jumlah_3011 = 0

for i_3011 in range(1, ulang_3011 + 1):
    print(i_3011, end = " ")
    jumlah_3011 += i_3011

    if i_3011 < ulang_3011:
        print("+", end = " ")
    else:
     print("=", jumlah_3011, end = " ")
print()
print ("Jumlah = ", jumlah_3011)
```

**Penjelasan :**
Variabel jumlah_3011 bertindak sebagai akumulator yang diawali 0. Pada setiap iterasi, nilai counter dicetak lalu ditambahkan ke jumlah_3011 dengan operator +=.
Percabangan if i_3011 < ulang_3011 mencetak tanda "+" selama iterasi belum mencapai angka terakhir. Saat iterasi terakhir, program mencetak tanda "=" beserta total penjumlahan, sehingga bentuk deret tampil rapi tanpa tanda "+" yang menggantung di ujung.
Setelah perulangan selesai, program menampilkan hasil akhir pada baris baru.


### 2.2.4 Program "jumlah_genap_2511533011"

```python
ulang_3011 = int(input("Masukkan jumlah perulangan: "))

jumlah_3011 = 0
for i_3011 in range(1, ulang_3011 + 1):
    if i_3011 % 2 == 0:
        print(i_3011, end = " ")
        jumlah_3011 += i_3011

        if i_3011 < ulang_3011:
            print("+", end = " ")
        else:
            print("=", jumlah_3011, end = " ")
print()
print("Jumlah = ", jumlah_3011)
```

**Penjelasan :**
Program serupa Program 3, tetapi hanya bilangan genap yang diproses. Syarat i_3011 % 2 == 0 memakai operator modulus: sebuah bilangan genap jika sisa baginya dengan 2 adalah 0. Perulangan tetap berjalan dari 1 sampai n, namun blok di dalam if hanya dieksekusi pada angka genap.
Akumulator jumlah_3011 menjumlahkan angka genap tersebut, dan tanda "+" dicetak setiap kali angka genap belum mencapai n.

Catatan hasil pengujian: tanda "=" hanya tercetak ketika n adalah bilangan genap, sebab pengecekan i_3011 < ulang_3011 berada di dalam blok if untuk bilangan genap, sedangkan iterasi terakhir (n yang ganjil) tidak pernah masuk ke blok tersebut. Akibatnya, untuk n = 7 deret berakhir dengan tanda "+" tanpa "=" (lihat contoh kedua). Total pada baris "Jumlah" tetap benar. Perbaikan yang mungkin adalah memeriksa i_3011 + 1 > ulang_3011 atau mencetak tanda "=" di luar perulangan.

### 2.2.5 Program "nested_for1_2511533011"

```python
batas_3011 = int(input("Masukkan nilai batas: "))
for line_3011 in range(1, batas_3011 + 1):
    for j_3011 in range(1, (-1 * line_3011 + batas_3011) + 1):
        print(".", end = " ")
    print(line_3011)
```

**Penjelasan :**
Program meminta nilai batas, lalu perulangan luar berjalan dari 1 sampai batas dan mewakili nomor baris (line_3011).
Pada setiap baris, perulangan dalam mencetak karakter "." sebanyak batas - line_3011 kali, ekspresi (-1 * line_3011 + batas_3011) setara dengan batas_3011 - line_3011. Setelah titik-titik tercetak, nomor baris dicetak lalu pindah ke baris berikutnya.
Akibatnya jumlah titik berkurang satu pada tiap baris dan baris terakhir hanya berisi angka, sehingga membentuk pola segitiga.


### 2.2.6 Program "nested_for2_2511533011"

```python
batas_3011 = int(input("Masukkan nilai batas: "))
for i_3011 in range(1, batas_3011 + 1):
    for j_3011 in range(1, batas_3011 + 1):
        print("*", end = " ")
    print()
```

**Penjelasan :**
Perulangan luar (i_3011) berjalan sebanyak batas kali dan mewakili jumlah baris. Perulangan dalam (j_3011) juga berjalan sebanyak batas kali dan mewakili jumlah kolom, dengan mencetak "*" yang dipisah spasi pada setiap iterasi.
print() tanpa argumen di akhir perulangan dalam memindahkan kursor ke baris berikutnya. Karena jumlah baris dan kolom sama, keluaran berbentuk persegi n x n.


### 2.2.7 Program "nested_for3_2511533011"

```python
batas_3011 = int(input("Masukkan nilai batas: "))
for i_3011 in range(batas_3011 + 1):
    for j_3011 in range(batas_3011 + 1):
        print(i_3011 + j_3011, end = " ")
    print()
```

**Penjelasan :**
Berbeda dengan Program 6, kedua perulangan memakai range(batas + 1) sehingga counter berjalan dari 0 sampai batas, dan terbentuk (n + 1) baris dan (n + 1) kolom.
Pada setiap sel, program mencetak hasil i_3011 + j_3011. Nilai di sel mana pun bertambah satu ke kanan dan satu ke bawah, sehingga tampak seperti tabel penjumlahan dengan nilai tertinggi 2n di sudut kanan bawah.


### 2.2.8 Program "nested_for4_2511533011"

```python
tinggi_3011 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_3011 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_3011 = tinggi_3011
    c_3011 = a_3011
    lebar_3011 = (2 * tinggi_3011) - 2

    for i_3011 in range(1, tinggi_3011 + 1):
        b_3011 = c_3011 + 1

        for j_3011 in range(1, lebar_3011 + 1):
            if i_3011 == 1 or i_3011 == tinggi_3011:
                if j_3011 == 1 or j_3011 == lebar_3011:
                    print("#", end = "")
                else:
                    print("=", end = "")

            else: 
                if j_3011 == 1 or j_3011 == lebar_3011:
                    print("|", end = "")
                else:
                    if j_3011 == c_3011:
                        print("<", end = "")
                    elif j_3011 == b_3011:
                        print(">", end = "")
                    elif j_3011 == (lebar_3011 - c_3011):
                        print("<", end = "")
                    elif j_3011 == (lebar_3011 - c_3011 + 1):
                        print(">", end = "")
                    elif j_3011 > b_3011 and j_3011 < (lebar_3011 - c_3011):
                        print(".", end = "")
                    else:
                        print(" ", end = "")
        print()

        # logika asli java
        a_3011 -= 2

        if a_3011 <= 0:
            c_3011 = (-a_3011) + 2
        else:
            c_3011 = a_3011

                    
```

**Penjelasan :**
Program meminta tinggi pola dan memvalidasinya: jika tinggi ganjil (tinggi % 2 != 0), program menampilkan pesan kesalahan dan berhenti. Jika genap, lebar pola dihitung dengan rumus (2 x tinggi) - 2.
Perulangan luar mengatur baris, perulangan dalam mengatur kolom. Baris pertama dan terakhir dicetak sebagai tepi atas dan bawah: "#" di kolom paling kiri dan kanan, "=" di antaranya. Pada baris lain, kolom paling kiri dan kanan dicetak "|".
Isi di antara kedua sisi ditentukan oleh variabel c_3011 dan b_3011 (b = c + 1). Tanda "<" dan ">" dicetak di kolom c dan b dari sisi kiri, serta di posisi cerminnya (lebar - c dan lebar - c + 1) 

### 2.2.9 Program "latihan5_2511533011"

```python
tinggi_3011 = int(input("Masukkan tinggi segitiga: "))

for i_3011 in range(1, tinggi_3011 + 1):
    for j_3011 in range(tinggi_3011 - i_3011):
        print(" ", end="")
    for k_3011 in range(i_3011):
        print("*", end=" ")
    print()
```

**Penjelasan :**
Perulangan luar berjalan dari 1 sampai tinggi dan mewakili nomor baris. Pada setiap baris, perulangan pertama mencetak spasi sebanyak (tinggi - i) sebagai indentasi, sehingga makin ke bawah spasi makin sedikit.
Perulangan kedua mencetak "*" sebanyak i kali, sehingga jumlah bintang bertambah satu pada tiap baris. Kombinasi spasi yang berkurang dan bintang yang bertambah membentuk segitiga sama kaki yang rata di bagian bawah.

Soal :
Studi Kasus: Pola Jam Pasir Kristal Palindromik Berbingkai (Framed Palindromic Crystal Hourglass)
Mahasiswa diminta merancang program konsol interaktif untuk mencetak pola bangun geometri dua dimensi berupa Jam Pasir Kristal Palindromik yang terbungkus rapi di dalam bingkai pembatas berpagar. Berbeda dengan pola dasar, program ini menuntut pembentukan deret angka yang bergerak dinamis menurun dan menaik secara palindromik pada setiap baris serta diselaraskan dengan poros kristal di pusat bangun.
Program menerima masukan nilai integer positif N dari pengguna yang merepresentasikan ukuran skala jam pasir (skala angka tertinggi pada puncak dan alas).

**Kode Jawaban :**
a. Kode Program "tugas5_2511533011" 

```python
print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")
n_3011 = int(input("Masukkan ukuran skala jam pasir (N): "))

print("#", end="")
for kolom_3011 in range(4 * n_3011 + 5):
    print("=", end="")
print("#", end="")
print()

for baris_3011 in range(n_3011, 0, -1):
    print("|", end="")
    print(" ", end="")
    for spasi_3011 in range(2 * (n_3011 - baris_3011)):
        print(" ", end="")
    for angka_3011 in range(baris_3011, 0, -1):
        print(angka_3011, end=" ")
    print("<*>", end="")
    for angka_3011 in range(1, baris_3011 + 1):
        print(" ", end="")
        print(angka_3011, end="")
    for spasi_3011 in range(2 * (n_3011 - baris_3011)):
        print(" ", end="")
    print(" ", end="")
    print("|", end="")
    print()

print("|", end="")
for spasi_3011 in range(2 * n_3011 + 1):
    print(" ", end="")
print("<*>", end="")
for spasi_3011 in range(2 * n_3011 + 1):
    print(" ", end="")
print("|", end="")
print()

for baris_3011 in range(1, n_3011 + 1):
    print("|", end="")
    print(" ", end="")
    for spasi_3011 in range(2 * (n_3011 - baris_3011)):
        print(" ", end="")
    for angka_3011 in range(baris_3011, 0, -1):
        print(angka_3011, end=" ")
    print("<*>", end="")
    for angka_3011 in range(1, baris_3011 + 1):
        print(" ", end="")
        print(angka_3011, end="")
    for spasi_3011 in range(2 * (n_3011 - baris_3011)):
        print(" ", end="")
    print(" ", end="")
    print("|", end="")
    print()

print("#", end="")
for kolom_3011 in range(4 * n_3011 + 5):
    print("=", end="")
print("#", end="")
print()
```

**Penjelasan :**

Program meminta pengguna memasukkan ukuran skala jam pasir (N) dan menyimpannya sebagai bilangan bulat pada variabel n_3011 lewat int(input(...)). Seluruh pola dibentuk hanya dengan perulangan for bersarang dan parameter end = "", tanpa perkalian string, dan setiap baris diakhiri print() tanpa argumen untuk pindah baris. Agar tidak menulis kode yang sama berulang, bagian bingkai dan baris berangka dibuat sebagai fungsi, yaitu border_3011() dan baris_angka_3011().

Bingkai atas dan bawah dicetak oleh border_3011(): karakter "#" di ujung kiri, perulangan range(4 * n_3011 + 5) yang mencetak "=" sebanyak 4N + 5 kali, lalu "#" di ujung kanan. Rumus 4N + 5 diperoleh dari lebar isi pola. Untuk N = 4, bagian tengah berisi 21 karakter "=", dan untuk N = 3 berisi 17 karakter.

Pola dibagi menjadi tiga fase. Fase 1 (jam pasir atas) memakai range(n_3011, 0, -1), sehingga baris_3011 turun dari N sampai 1. Fase 3 (jam pasir bawah) memakai range(1, n_3011 + 1), sehingga baris_3011 naik dari 1 sampai N. Keduanya memanggil fungsi baris_angka_3011() yang sama, sehingga bagian atas dan bawah simetris. Fase 2 adalah poros pusat yang dicetak satu kali di antara kedua fase.
